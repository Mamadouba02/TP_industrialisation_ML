# Exploration du churn
Données synthétiques pédagogiques ; exécuter les cellules depuis la racine ou notebooks.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
root = Path.cwd() if (Path.cwd() / 'data/raw/churn.csv').exists() else Path.cwd().parent
df = pd.read_csv(root / 'data/raw/churn.csv')
df.head()
```

```python
print(df.shape)
print(df.dtypes)
print(df.isna().sum())
df.describe()
```

```python
print(df.churn.value_counts(normalize=True))
sns.countplot(data=df, x='churn')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, fmt='.2f')
plt.show()
```

## Interprétation
La formule génératrice favorise le churn quand le montant et les retards augmentent et quand l’ancienneté diminue. Les corrélations ne prouvent pas une causalité. Les résultats sur ce jeu synthétique ne garantissent pas une généralisation sur des clients réels.
